```python
# prompt: a plot using the subplots library with data stored in a pandas dataframe and the font STIX for all labels and ticks

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# KartheinLab color scheme
cols = ['#FFCC00', '#FF2D55', '#00A2FF', '#61D935', 'k', 'grey'] # yellow, red, blue, green, black, gray

# Sample DataFrame
df = pd.DataFrame({
    'x': np.linspace(0, 10, 100),
    'y': np.sin(np.linspace(0, 10, 100)),
    'yerr': np.sqrt(np.abs(np.sin(np.linspace(0, 10, 100))))
})

# Use STIX font for all labels and ticks
plt.rcParams['font.family'] = 'STIXGeneral'
plt.rcParams['mathtext.fontset'] = 'stix'

# Create a figure
fig, ax = plt.subplots(1,1, figsize=(6,4))

# Plot the data
ax.errorbar(df.x, df.y, df.yerr, color='#FF2D55')

# Set labels and font properties
ax.set_xlabel('X-Axis (Unit)', fontsize=15, fontweight='bold')
ax.set_ylabel('Y-Axis (Unit)', fontsize=15, fontweight='bold')
ax.tick_params(axis='both', which='major', labelsize=13)

# Save the plot
plt.savefig('test_plot.pdf', bbox_inches='tight', pad_inches=0.1) # please always save the plot as a vector graphic
plt.savefig('test_plot.png', dpi=300, bbox_inches='tight', pad_inches=0.1) # 300 DPI if necessary

# Show the plot
plt.show()
```
