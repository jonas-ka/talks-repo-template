**Karthein Lab figure template.** Make plots in Google Colab, one folder per figure in the Shared Drive's `Figures/` folder (its `README.md` has the rules).

1. Run the first cell: it loads the lab style and the figure helper `labfig` from the public lab-templates repository.
2. Draw the figure with `plt.subplots` (6 x 4 in, 15 pt bold axis labels, 13 pt ticks, the lab colours), then save it with `labfig.save(...)`: PDF, SVG and a 300 dpi PNG from the same figure, into the figure's folder, named `<name>-v<N>`. **A version is never overwritten:** every visual change is the next version; while you are still working, save with `draft=True`.
3. Below the figure, `labfig.save` adds an **information strip**: who made it and when, who changed it last and when, the template version (linked), and the Drive folder (linked in PDF and SVG). **Cut the strip off** when the figure goes into a paper, talk or poster: it is the bottom 0.4 in, under the dashed line. The lab's Typst themes cut it automatically.
4. Paste the figure folder's link as `drive_url` (Share → Copy link), so the PDF and SVG link back to the folder.

To change a figure (its look or its data), save the next version beside the old ones, with the original `author` and `created`, plus `modified_by` and `modified`; changed data goes into `data/` under a new name. Nothing in a figure folder is overwritten or moved (see the README). Make use of Google Gemini when coding 😏

```python
# The lab style and the figure helper, from the public lab-templates repository (the latest
# version; for a fixed one, add  --branch v2026.10.10  to the clone).
import os, sys
if not os.path.isdir("/content/lab-templates"):
    !git clone -q --depth 1 https://github.com/jonas-ka/lab-templates.git /content/lab-templates
sys.path.insert(0, "/content/lab-templates/matplotlib")
import labfig
labfig.style()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Sample data in a DataFrame
x = np.linspace(0, 10, 30)
df = pd.DataFrame({"x": x, "y": np.sin(x), "yerr": 0.1 + 0.1 * np.abs(np.sin(x))})

# One figure, drawn with plt.subplots; the style sets size, fonts, colours and the frame.
# Lab colours: #FFCC00 yellow, #FF2D55 red, #00A2FF blue, #61D935 green (one colour per thing).
fig, ax = plt.subplots()
ax.errorbar(df.x, df.y, df.yerr, fmt="o-", color="#FF2D55", label="Measurement")
ax.set_xlabel("Time (s)")
ax.set_ylabel("Signal (V)")
ax.legend(loc="lower left")   # inside the box, no frame, where there is no data

name = "plot-example-sine"    # the figure's folder and file name
# Saves plot-example-sine-v1.pdf/.svg/.png. A version is never overwritten: every visual change
# is the next version (version=2, with modified_by and modified). While working: draft=True.
# Also writes plot-example-sine-v1.yaml (the record: who, when, template, data fingerprint).
labfig.save(fig, name, version=1, author="Your Name", created="2026-10-13",
            data=None,         # the data files the figure draws, e.g. [f"data/{name}.csv"]: fingerprint on the strip
            code=None,         # the path of this notebook on the mounted Drive: a copy goes to code/
            drive_folder=f"Figures/{name}",
            drive_url=None)    # paste the folder's link here (Share -> Copy link)
plt.show()
```
